## 小型天和 smoke run

先用較小樣本確認 Numba compilation 與核心函式可以正常執行，並非為正式研究結果。


```python
weights = PRESET_NORMAL.astype(np.float64)

estimate, hits = simulate_tenhou_batch(
    n=100000,
    event_code=EVENT_TOTAL,
    weights=weights,
    seed=42,
)

print("Tenhou estimate (small smoke run):", estimate)
print("Proposal hits:", hits)
print("Benchmark:", TENHOU_BENCHMARKS[EVENT_TOTAL])
```

## 小型地和 smoke run

以下以南家、普通型、`p_call=0.5` 做小型執行檢查。


```python
weighted, raw = simulate_chiho_batch(
    n=100000,
    target_seat=1,
    event_code=EVENT_NORMAL,
    weights=weights,
    p_chi=0.5,
    seed=42,
)

print("Chiihou weighted diagnostics:", weighted)
print("Chiihou raw counts:", raw)
```

## 在本機啟動 Streamlit

執行下一個 chunk 後，Streamlit 會直接執行 repository root 中的 `app.py`。

將會直接開啟 Streamlit 的連結即可操作。


```python
import sys
import subprocess
import time
import webbrowser
from pathlib import Path

app_path = Path("app.py").resolve()

process = subprocess.Popen(
    [
        sys.executable,
        "-m",
        "streamlit",
        "run",
        str(app_path),
        "--server.port",
        "8501",
    ]
)

time.sleep(3)

print("Streamlit process PID:", process.pid)
print("Local URL: http://localhost:8501")

webbrowser.open("http://localhost:8501")
```
